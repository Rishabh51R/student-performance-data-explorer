# Student Performance Data Explorer

## AI/ML Internship — Task 1

This notebook performs exploratory data analysis (EDA) on the provided student performance dataset using **Python, Pandas, NumPy, Matplotlib, and Seaborn**.

### Workflow
1. Load the CSV dataset
2. Inspect the data
3. Check missing values and duplicates
4. Clean the dataset
5. Calculate descriptive statistics
6. Calculate mean, median, and mode
7. Create histogram, scatter plot, box plots, and correlation heatmap
8. Summarize the main findings

> **Note:** Correlation indicates association; it does not by itself prove causation.

## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## 2. Load the Dataset

```python
# Keep the CSV file in the same folder as this notebook.
df = pd.read_csv("student_performance.csv")

print("Dataset loaded successfully.")
```

## 3. Display the First 10 Rows

```python
df.head(10)
```

## 4. Dataset Shape and Columns

```python
print("Rows and columns:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

## 5. Data Types

```python
df.dtypes
```

## 6. Missing-Value Analysis

```python
missing_values = df.isnull().sum()
print(missing_values)

print("\nTotal missing cells:", int(missing_values.sum()))
```

## 7. Duplicate Detection and Cleaning

```python
duplicate_count = int(df.duplicated().sum())
print("Duplicate rows before cleaning:", duplicate_count)

# Remove duplicate rows, if any.
df_clean = df.drop_duplicates().copy()

# Fill missing numeric values with the median of each numeric column, if any exist.
numeric_columns = df_clean.select_dtypes(include=np.number).columns
for col in numeric_columns:
    if df_clean[col].isnull().any():
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())

print("Shape after cleaning:", df_clean.shape)
print("Missing cells after cleaning:", int(df_clean.isnull().sum().sum()))
print("Duplicate rows after cleaning:", int(df_clean.duplicated().sum()))
```

## 8. Descriptive Statistics

```python
df_clean.describe()
```

## 9. Mean, Median, and Mode

```python
numeric_df = df_clean.select_dtypes(include=np.number)

summary_stats = pd.DataFrame({
    "Mean": numeric_df.mean(),
    "Median": numeric_df.median(),
    "Mode": numeric_df.mode().iloc[0]
})

summary_stats
```

## 10. Final Score Distribution — Histogram

```python
plt.figure(figsize=(8, 5))
sns.histplot(df_clean["final_score"], bins=8, kde=True)
plt.title("Final Score Distribution")
plt.xlabel("Final Score")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

## 11. Study Hours vs Final Score — Scatter Plot

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df_clean,
    x="study_hours",
    y="final_score",
    hue="gender",
    s=80
)
plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.tight_layout()
plt.show()
```

## 12. Numeric Variables — Box Plots

```python
box_columns = [
    "study_hours",
    "attendance_percent",
    "previous_score",
    "assignments_completed",
    "final_score"
]

plt.figure(figsize=(10, 5))
sns.boxplot(data=df_clean[box_columns])
plt.title("Box Plots of Numeric Variables")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## 13. Correlation Matrix

```python
correlation_matrix = numeric_df.corr()
correlation_matrix
```

## 14. Correlation Heatmap

```python
plt.figure(figsize=(9, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## 15. Key Findings

```python
avg_final = df_clean["final_score"].mean()
median_final = df_clean["final_score"].median()
highest_final = df_clean["final_score"].max()
lowest_final = df_clean["final_score"].min()
avg_study = df_clean["study_hours"].mean()
study_final_corr = df_clean["study_hours"].corr(df_clean["final_score"])

final_correlations = correlation_matrix["final_score"].drop("final_score").sort_values(
    key=lambda s: s.abs(), ascending=False
)

print(f"Average final score: {avg_final:.2f}")
print(f"Median final score: {median_final:.2f}")
print(f"Highest final score: {highest_final:.2f}")
print(f"Lowest final score: {lowest_final:.2f}")
print(f"Average study hours: {avg_study:.2f}")
print(f"Study hours / final score correlation: {study_final_corr:.2f}")
print("\nCorrelations with final score:")
print(final_correlations)
```
